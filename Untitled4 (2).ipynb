```python
X = df.drop(columns=['Channel', 'Region'])
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

```python
scaled_df = pd.DataFrame(X_scaled, columns=X.columns)

scaled_df.head()
```

```python
from sklearn.cluster import KMeans

wcss = []

for i in range(1,11):
    kmeans = KMeans(n_clusters=i, random_state=42)
    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)
```

```python
plt.figure(figsize=(8,5))
plt.plot(range(1,11), wcss, marker='o')
plt.title("Elbow Method")
plt.xlabel("Number of Clusters")
plt.ylabel("WCSS")
plt.show()
```

```python
kmeans = KMeans(n_clusters=5, random_state=42)
clusters = kmeans.fit_predict(X_scaled)
```

```python
df["Cluster"] = clusters
df.head()
```

```python
df["Cluster"].value_counts()
```

```python
df.groupby("Cluster").mean()
```

```python
from sklearn.decomposition import PCA
pca = PCA(n_components=2)

X_pca = pca.fit_transform(X_scaled)
X_pca.shape
```

```python
plt.figure(figsize=(8,6))

plt.scatter(
    X_pca[:,0],
    X_pca[:,1],
    c=clusters,
    cmap='viridis'
)

plt.title("Customer Segmentation using K-Means")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")

plt.colorbar(label="Cluster")

plt.show()
```

```python
from scipy.cluster.hierarchy import dendrogram, linkage
linked = linkage(X_scaled, method='ward')
```

```python
plt.figure(figsize=(12,6))

dendrogram(linked)

plt.title("Dendrogram")
plt.xlabel("Customers")
plt.ylabel("Euclidean Distance")

plt.show()
```

```python
from sklearn.cluster import AgglomerativeClustering

hc = AgglomerativeClustering(
    n_clusters=5,
    metric='euclidean',
    linkage='ward'
)

hc_clusters = hc.fit_predict(X_scaled)
```

```python
df["HC_Cluster"] = hc_clusters
df["HC_Cluster"].value_counts()
```

```python
df.groupby("HC_Cluster").mean()
```

```python
from sklearn.cluster import DBSCAN
dbscan = DBSCAN(eps=1.5, min_samples=5)

db_clusters = dbscan.fit_predict(X_scaled)
df["DBSCAN_Cluster"] = db_clusters
df["DBSCAN_Cluster"].value_counts()
```

```python
df.groupby("DBSCAN_Cluster").mean()
```

```python
plt.figure(figsize=(8,6))

plt.scatter(
    X_pca[:,0],
    X_pca[:,1],
    c=db_clusters,
    cmap='rainbow'
)

plt.title("DBSCAN Clustering")

plt.xlabel("PC1")
plt.ylabel("PC2")

plt.colorbar(label="Cluster")

plt.show()
```

```python
from sklearn.metrics import silhouette_score

kmeans_score = silhouette_score(X_scaled, clusters)

print("K-Means Silhouette Score:", kmeans_score)
```

```python
hc_score = silhouette_score(X_scaled, hc_clusters)

print("Hierarchical Silhouette Score:", hc_score)
```

```python
if len(set(db_clusters)) > 1:
    dbscan_score = silhouette_score(X_scaled, db_clusters)
    print("DBSCAN Silhouette Score:", dbscan_score)
else:
    print("DBSCAN could not form enough clusters.")
```

```python
results = pd.DataFrame({
    "Algorithm": ["K-Means", "Hierarchical", "DBSCAN"],
    "Silhouette Score": [
        kmeans_score,
        hc_score,
        dbscan_score if len(set(db_clusters)) > 1 else None
    ]
})

results
```

```python
results.sort_values(by="Silhouette Score", ascending=False)
```

```python
print(kmeans_score)
print(hc_score)
print(dbscan_score)
```

# Conclusion
Three clustering algorithms (K-Means, Hierarchical Clustering, and DBSCAN) were applied to the Wholesale Customers dataset. Data preprocessing included exploratory data analysis, feature scaling using StandardScaler, and visualization using PCA. The models were evaluated using the Silhouette Score. The results showed that DBSCAN achieved the highest score (0.6602), outperforming K-Means (0.3689) and Hierarchical Clustering (0.2399). Therefore, DBSCAN was the most effective algorithm for customer segmentation on this dataset because it produced more compact and well-separated clusters.

```python
%%writefile app.py

import streamlit as st
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA

st.title("Customer Segmentation")

df = pd.read_csv("Wholesale customers data.csv")

st.write(df.head())

X = df.drop(columns=["Channel","Region"])

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

k = st.slider("Number of Clusters",2,10,5)

kmeans = KMeans(n_clusters=k, random_state=42)
clusters = kmeans.fit_predict(X_scaled)

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

fig, ax = plt.subplots()

ax.scatter(X_pca[:,0], X_pca[:,1], c=clusters)

st.pyplot(fig)
```
